```python
from ultralytics import YOLO
```

```python
# Load a pretrained object detection model
model = YOLO("yolo26n.pt")
```

```python
# Run inference on an example image
results = model.predict(
    source="bus.jpg",
    save=True,
    conf=0.25,
    # device="mps"
)

# Examine the model's output
for result in results:
    for box in result.boxes:
        print("Class:", result.names[int(box.cls.item())])
        print("Confidence:", box.conf.item())
        print("Bounding box:", box.xyxy.tolist())
        print("---")
```
