```python
import torch
import numpy as np
```

```python
data = [[1,2],[3,4]]
x_data = torch.tensor(data)
```

```python
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
```

```python
#preserver the shape but changes the values

x_ones = torch.ones_like(x_data)
print(f"One Tensor:\n {x_ones}\n")

x_rand = torch.rand_like(x_data, dtype=torch.float)
print(f"Random Tensor: \n {x_rand} \n")
```

```python
shape = (2,3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)
```

```python

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```

```python
tensor = torch.rand(3,4)

print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

By default, tensors are created on the CPU. We need to explicitly move tensors to the accelerator using .to method (after checking for accelerator availability). Keep in mind that copying large tensors across devices can be expensive in terms of time and memory!

```python
# We move our tensor to the current accelerator if available
if torch.accelerator.is_available():
    print(torch.accelerator.current_accelerator())
    tensor = tensor.to(torch.accelerator.current_accelerator())
```

```python
tensor = torch.ones(4, 4)
print(f"First row: {tensor[0]}")
print(f"First column: {tensor[:, 0]}")
print(f"Last column: {tensor[..., -1]}")
tensor[:,1] = 0
print(tensor)
```

```python
t0 = torch.cat([tensor, tensor, tensor], dim=0)
print(t0)
```

```python
t1 = torch.cat([tensor, tensor, tensor], dim=1)
print(t1)
```

```python
agg = tensor.sum()
agg_item = agg.item()
print(agg)
print(agg_item, type(agg_item))
```

```python
print(f"{tensor} \n")
tensor.add_(5)
print(tensor)
```
